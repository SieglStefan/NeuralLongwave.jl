## **0. Setup**

```julia
# Load packages
using Revise
using NeuralParam
```

```julia
# Define experiment and series
EXP = "exp_B"
SER = "02_NLW_MLP_mixed_pilot";

# Define output forms and colors
OUTPUT_FORMS  = ["direct", "linear", "planck"];
COLORS_OF = [JL_BLUE, JL_GREEN, JL_RED];
```

## **1. Training**

### 1.1. Zero Eta Baseline
