# **Setup OBLW-05: Reference Runs**

Sanity check of the references every emulator is evaluated against: the climate noise floor against the reference climate, and the grey OneBandLongwave (constant transmissivity) as a simpler physical scheme a good emulator has to beat.

Generated by `scripts/targets/OBLW/03_reference.jl` (`rollout` tasks 0-3) → `results/OBLW/{climate_ref, climate_noise, grey}/`

## **0. Setup**

```julia
# Load packages
using Revise
using NeuralLongwave
```

```julia
# Reference climate, climate noise floor, grey baseline (weather and climate) and the RMSE caps
ref          = load_reference("OBLW", "climate_ref",   "climate.jld2")
noise        = load_reference("OBLW", "climate_noise", "climate.jld2")
grey_weather = load_reference("OBLW", "grey",          "weather.jld2")
grey_climate = load_reference("OBLW", "grey",          "climate.jld2")
caps         = load_reference("OBLW", "spinup",        "rmse_caps.jld2");
```

## **1. Climate Noise Floor**

```julia
# Drift of the noise floor against the reference climate (must stay around zero)
plot_climate_drift((; climate_noise = noise), ref)
```

```julia
# Global bias with its 95 % confidence interval (must contain zero)
plot_climate_bias((; climate_noise = noise), ref)
```

```julia
# Zonal pattern of the noise floor (what chance alone produces with this ensemble)
plot_climate_zonal((; climate_noise = noise), ref, :T)
```

## **2. Grey Baseline**

```julia
# Weather: RMSE growth of the grey scheme, with the RMSE cap (zero skill) for comparison
plot_weather_growth((; grey = grey_weather), :rmse; caps)
```

```julia
# 14-day RMSE of the grey scheme: the reference line of the weather-cost plane (ref_rmse)
ref_rmse = weather_table((; grey = grey_weather)).weather_rmse[1]
```

```julia
# Climate: bias of the grey scheme next to the noise floor
plot_climate_bias((; climate_noise = noise, grey = grey_climate), ref)
```

```julia
# Zonal bias of the grey scheme next to the noise floor
plot_climate_zonal((; climate_noise = noise, grey = grey_climate), ref, :T)
```
