# **Experiment A-04 - Dependence of Trajectory Length**

---
## **🔴 Context**

- Until now, for the parameter updates, a trajectory of `n_steps = 10` length was being differentiated.
- Differentiation of longer trajectories is computationally more expensive, but may yield better results in e.g. stability.
- The distance between two starts of a differentiated trajectory was held constant (`n_steps + n_gap = 60`).

---

## **🟡 Experiment Questions**


#### **Q1 - Do shorter trajectories yield better/same results as the default?**

No, none of the tested short trajectories (S1,S2,S5) yielded better results. S1 and S2 perform very similar, S5 is closer to S10.


#### **Q2 - Do longer trajectories yield better results?**

S15 and S20 perform slightly worse than the default S10. S30 performs significantly worse than S10.


---


## **🟢 Short Result**
#### The default step size of `n_steps = 10` performs best. But this could be coupled to the gap size `n_gap`.
